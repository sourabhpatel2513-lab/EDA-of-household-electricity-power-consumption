```python
num_data = ['Global_active_power', 'Global_reactive_power',
       'Voltage', 'Global_intensity', 'Sub_metering_1', 'Sub_metering_2',
       'Sub_metering_3']
```

#### coverting the date column into datetime

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df.head()
```

### extracting  day_name, month_name , year  from data


```python
df["day"] = df["Date"].dt.day
```

```python
df["day_name"] = df["Date"].dt.day_name()
```

```python
df["year"] = df["Date"].dt.year
```

```python
df["month_name"] = df["Date"].dt.month_name()
```

```python
df.head()
```

```python
df.drop(columns = ["Date"])
```
