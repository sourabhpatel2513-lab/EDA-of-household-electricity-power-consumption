```python
import matplotlib.pyplot as plt
import seaborn as sns
pos=1

plt.figure(figsize=(14, 10))  # Increased size slightly so 45-degree labels don't crowd
cat_2 = ['year', 'month_name', 'day_name', 'Hours']

for i in cat_2:
    
    ax = plt.subplot(2, 2, pos)
    pos=pos+1
    
    # 2. Prepare data
    grouped = df.groupby(i)['Total_metering'].sum()
    x = grouped.index
    y = grouped.values
    
    
    sns.barplot(x=x, y=y, ax=ax)
    
    ax.bar_label(ax.containers[0], padding=9, fontsize=9)
    
    plt.xticks(rotation=45)
    ax.set_ylim(0, max(y) * 1.15) 
    ax.set_title(f'Total Metering by {i.capitalize()}')

plt.tight_layout()
plt.show()
```
